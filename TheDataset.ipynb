# The Cerca/QuSpin data set 


## Introduction 

To demonstrate the functionality of the pipeline, we will be using MEG datasets recorded using SQUID- and OPM-based MEG data employing a spatial attention paradigm. The paradigm is derived from a study involving moving target gratings and the detection of subtle dots (based on [1,2]; see Figure 1). The main advantage of this task is that the allocation of spatial attention produces robust modulations in the 8 – 12 Hz alpha band during the cue-target interval [3]. Furthermore, the moving gratings will induce strong activity in the 60 – 90 Hz gamma band, also modulated by attention. 


## The OPM-MEG system 

The data were recorded using the OPM-MEG system from Cerca using sensors from QuSpin installed 2025 at OHBA. This system has 64 sensors distributed over the head. Each sensor is triaxial, meaning that it will measure the magnetic field in 3 directions. One of these directions corresponds to conventional magnetometers measuring field perpendicular to the scalp surface.

The OPM-MEG system is placed in a shielded room (2 layers of mu-metal and 1 layer of aluminium).  

The peripherals of the OPM-MEG system applied to record the dataset include:

* ProPixx visual projector (1440 Hz refresh rate)
* Optical button boxes
* Stimulus delivery computer using PsychoPy.

The trigger were controlled by the stimulus computer communicating with the DataPixx. The DataPixx then sent triggers via a parallel port to the Cerca aquisition system.

Please see "The Introduction" for where to find the data.



## The paradigm 
Trials started with a fixation dot followed by a cue pointing to the left or right (Figure 1). This indicated to the participants which of the two upcoming stimuli to attend. After a 1 s interval, two inward-moving circular gratings were shown in both hemifields. After a random interval (0.1s – 1.9s) a white dot was presented for 50 ms at the centre of the grating in the cued hemifield. Gratings remained on the scren for a fixed duration of 3s. Participants were instructed to press a button (right index finger) as soon as they detected the dot. The stimuli presentation were implemented in Python using the PsychoPy and the code can be found on the FLUX website.

**Figure 1.** *The paradigm used for the OPM-MEG data collection in the FLUX pipeline. A cue directs the participants to attend left or right. This results in strong hemispheric lateralisation of the posterior alpha power (8 – 12 Hz). Then follows the presentation of the moving gratings, which will induce strong gamma power (60 – 90 Hz) in the visual cortex as well as visual evoked fields. Finally, the button response to the dot flash will result in gamma band activity in the motor cortex, followed by suppression in the beta band (15 – 40 Hz). Since the task produces robust and well-documented modulations in the alpha, beta and gamma band as well as event-related fields, it is well suited for the FLUX pipeline with a focus on the characterisation of oscillatory brain activity in the visual and motor regions as well as source modelling.*

Publication example. Describe the relevant parameters of the tasks: 

"Each trial started with a fixation dot followed by a cue pointing to the left or right (Figure 1). This indicated to the participants which of the two upcoming stimuli to attend to. After a 1 s interval, two inward-moving circular gratings were shown in both hemifields. The cue was randomly to the left or the right. After a random interval (0.2 – 2.0 s), a white dot was presented for 50 ms at the centre of the grating in the cued hemifield. The gratings always remained on for 2.0 s. Participants were instructed to press a button (right index finger) as soon as they detected the dot. The moving gratings were centred at +/- 4 degrees. They were each 2 degrees wide. In total 100 trials were presented to each participant. The stimulus presentation was implemented in PsychoPy.  "
